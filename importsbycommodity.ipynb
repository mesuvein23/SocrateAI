```python
china_total_import_value_percent = (china_total_import_value / df['Imports_Value'].sum()) * 100
print(f"The percent of china total import value is:{china_total_import_value_percent:.2f}%")

india_total_import_value_percent = (india_total_import_value / df['Imports_Value'].sum()) * 100
print(f"The total percent of india total import value is: {india_total_import_value_percent:.2f}%")
```

```python
df2 = df2[df2['HSCode'].notna()] # Keep only rows where HSCode is not blank (NaN).
total_value = df2['Imports_Value'].sum() # Calculate the grand total of all import values (across every commodity & country).

# Group the data by HSCode + Description (i.e., by unique commodity),
# then sum the Imports_Value for each commodity across all partner countries,
# then sort so the highest-value commodities appear first.
commodity_share = df2.groupby(['HSCode', 'Description'])['Imports_Value'].sum().sort_values(ascending=False)
commodity_share_pct = (commodity_share / total_value * 100).round(2)
print(commodity_share_pct)
```
